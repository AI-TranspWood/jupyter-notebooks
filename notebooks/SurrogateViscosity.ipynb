```python
import os
import ipywidgets as wdg
import utilities
from common import VISCOSITY_SURROGATE_M1_INF_DEFAULTS, VISCOSITY_SURROGATE_M1_TRAIN_DEFAULTS, VISCOSITY_SURROGATE_OUTDIR
from aitw_viscosity_surrogate.cli.xgboost_baseline import train, inference

m1_outdir = os.path.join(VISCOSITY_SURROGATE_OUTDIR, 'm1')
m1_train_outdir = os.path.join(m1_outdir, 'train')

widget_m1, m1_event = utilities.click_cli_widget(
    inference,
    override_defaults=VISCOSITY_SURROGATE_M1_INF_DEFAULTS,
    title='INFERENCE',
    button_description='Evaluate',
    button_running_text='Evaluating...',
    output_file_param='output_file',
)

m1_visualizer = utilities.output_visualizer_widget(
    title="Inference Results",
    file_format="csv",
    event_observer=m1_event,
    output_dir=m1_outdir,
)
visualizer_accordion = wdg.Accordion(children=[m1_visualizer], selected_index=None)
visualizer_accordion.set_title(0, 'Inference Results')

widget_m1_train, m1_train_event = utilities.click_cli_widget(
    train,
    override_defaults=VISCOSITY_SURROGATE_M1_TRAIN_DEFAULTS,
    title='TRAIN',
    button_description='Train',
    button_running_text='Training...',
    output_dir_param='output_dir'
)
train_m1_visualizer = utilities.output_visualizer_widget(
    title="Training Results",
    file_format="png",
    event_observer=m1_train_event,
    output_dir=m1_train_outdir,
)
train_m1_visualizer_accordion = wdg.Accordion(children=[train_m1_visualizer], selected_index=None)
train_m1_visualizer_accordion.set_title(0, 'Training Results')
train_m1_box = wdg.VBox(children=[widget_m1_train, train_m1_visualizer_accordion])
train_m1_accordion = wdg.Accordion(children=[train_m1_box], selected_index=None)
train_m1_accordion.set_title(0, 'Train Model')

m1_vbox = wdg.VBox(children=[widget_m1, visualizer_accordion, train_m1_accordion])
m1_accordion = wdg.Accordion(children=[m1_vbox], selected_index=None)
m1_accordion.set_title(0, 'Method 1: XGBoost with Molecular Fingerprints and Descriptors')
display(m1_accordion)
```
