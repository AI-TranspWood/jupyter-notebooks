```python
import os

def performance_finder(base_path: str) -> list[tuple[str, str]]:
    """Find performance files in the base path.

    Args:
        base_path (str): The base directory to search for performance files.

    Returns:
        list[tuple[str, str]]: A list of tuples containing the label and file path of performance files.
    """
    for root, dirs, files in os.walk(base_path):
        for file in files:
            if file.endswith("_performance.png") or file.endswith("_predictions.png"):
                dirname = os.path.basename(root)
                filename_without_ext = os.path.splitext(file)[0]
                label = f"{dirname}: {filename_without_ext}"
                yield (label, os.path.join(root, file))
```
